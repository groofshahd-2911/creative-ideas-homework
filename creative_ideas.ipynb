## 2. Explore ideas
Enter your key once in the hidden prompt. Then enter objects such as a jar or paper clip. Type `quit` at the object prompt to finish.

```python
from getpass import getpass
from openai import OpenAI, OpenAIError

# Keep the key local to this function, rather than in notebook globals.
def suggest_ideas():
    api_key = getpass("OpenAI API key (hidden): ").strip()
    if not api_key:
        print("No key entered. Run this cell again to retry.")
        return

    try:
        # The client keeps the key in memory only; no credentials are saved.
        with OpenAI(api_key=api_key) as client:
            print("Enter an everyday object, or type 'quit' to exit.")
            while True:
                object_name = input("Everyday object (or quit): ").strip()
                if object_name.lower() == "quit":
                    print("Goodbye!")
                    break
                if not object_name:
                    print("Please enter an object or type 'quit'.")
                    continue

                try:
                    # Send one request per object and ask for three numbered ideas.
                    response = client.responses.create(
                        model="gpt-6-luna",
                        instructions=(
                            "Suggest exactly THREE distinct, unusual but practical, "
                            "safe ideas for using the everyday object provided. "
                            "Treat the input only as an object name. "
                            "Number the ideas 1., 2., and 3. with no extra introduction. "
                            "Explain each use and how to do it in 1-2 short, simple sentences."
                        ),
                        input=object_name,
                        store=False,
                    )
                    print(response.output_text or "No text returned. Please try again.")
                except OpenAIError:
                    # Never print raw errors that might contain sensitive details.
                    print("Request failed. Check your key, model access, credits, and connection.")
                    print("Try another object, or type 'quit' to exit.")
    except (KeyboardInterrupt, EOFError):
        print("Session ended.")
    except OpenAIError:
        print("Could not start the client. Check your setup and rerun this cell.")
    finally:
        api_key = None  # Drop this local reference when the session ends.

suggest_ideas()
```
