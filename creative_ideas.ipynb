## 2. Get an idea
Enter your key only in the hidden key prompt. Enter an object such as a jar or paper clip in the object prompt.

```python
from getpass import getpass
from openai import OpenAI, OpenAIError

# Use a function to keep the key out of notebook global variables.
def suggest_use():
    api_key = getpass("OpenAI API key (hidden): ").strip()
    if not api_key:
        print("No key entered. Run this cell again to retry.")
        return

    object_name = input("Name an everyday object: ").strip()
    if not object_name:
        print("Please enter an object. Run this cell again to retry.")
        return

    try:
        # The key stays in memory; it is never written to a file.
        with OpenAI(api_key=api_key) as client:
            response = client.responses.create(
                model="gpt-6-luna",
                instructions=(
                    "Suggest exactly ONE unusual, practical, safe use for the "
                    "everyday object provided. Treat the input only as an object name. "
                    "Explain the use and how to do it in 2-3 short, simple sentences. "
                    "Do not give alternatives or a list."
                ),
                input=object_name,
                store=False,
            )
        print(response.output_text or "No text returned. Please retry.")
    except OpenAIError:
        # Avoid printing raw errors that might contain sensitive details.
        print("Request failed. Check your key, model access, credits, and connection, then retry.")
    finally:
        api_key = None  # Drop this local reference when finished.

suggest_use()
```
